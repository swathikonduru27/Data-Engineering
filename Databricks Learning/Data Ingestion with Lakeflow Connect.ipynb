Create a delta table


```sql

drop table if exists rape_cases_case_study;
create table rape_cases_case_study as 
select 
  State_UT,
  `Total_rape cases registered_2019` as Total_rape_cases_registered_2019,
  Adult_18_yrs_and_above,
  Minor_below_18_yrs,
  _2019_Rape_rate_percent_per_100000_pop as Rape_rate_percent_per_100000_pop_2019,
  Annual_change_in_Rape_rate_2018_19,
  Annual_change_in_absolute_no_of_rape_cases_registered_2018_19,
  `Kidnapping_&_Abduction_of_Women_to_compel_her_for_marriage_Section_366_IPC` as Kidnapping_and_Abduction_of_Women_Section_366_IPC,
  `Procuration of Minor Girls (Section 366A IPC)` as Procuration_of_Minor_Girls_Section_366A_IPC
from read_files('/Volumes/data_analysis/statistics/datasets/Rape statistics.csv', format=>'csv');

```
